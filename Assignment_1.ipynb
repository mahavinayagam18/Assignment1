Question 4: An e-commerce company wants an AI system that:

● Automatically places supplier orders
● Negotiates price via email
Explain:
1. Why this is an example of Agentic AI.
2. What risks might arise?
3. What human oversight would still be required?

1. Why is this an example of Agentic AI?
This is an example of Agentic AI because the system can work independently. It monitors inventory, predicts demand, places orders with suppliers, and even negotiates prices through email without needing human involvement at every step.

2. What risks might arise?
The AI may make wrong predictions, order too much or too little stock, agree to a high price, or send incorrect messages to suppliers. There could also be security and privacy issues.

3. What human oversight is required?
Humans are still needed to monitor the system and make sure it is working correctly. They should approve large orders, set limits on prices and order quantities, and handle important or unusual situations.

Question 6: A company gives bonus based on performance score:
● Score ≥ 90 : 20% bonus
● Score ≥ 75 : 10% bonus
● Else : No bonus
Write Python logic using:
● Comparison operators
● Logical operators
● Conditional statements

```python
score = int(input("Enter your performance score: "))
if score >= 90:
    bonus = 20
elif score >= 75 and score < 90:
    bonus = 10
else:
    bonus = 0

print("Bonus:", bonus, "%")
```

Question 7: You are given a customer review:
review = "The product quality is Excellent but delivery was Late"
Write Python code to:
1. Convert to lowercase
2. Count number of words
3. Check if the word "excellent" exists
4. Replace "Late" with "delayed"

```python
review = "The product quality is Excellent but delivery was Late"

# 1. Convert to lowercase
lower_review = review.lower()
print(lower_review)

# 2. Count number of words
word_count = len(review.split())
print("Number of words:", word_count)

# 3. Check if "excellent" exists
print("excellent" in lower_review)

# 4. Replace "Late" with "delayed"
new_review = review.replace("Late", "delayed")
print(new_review)
```

Question 8: Given sales data:
sales = [12000, 15000, 11000, 18000, 17000]
1. Calculate total revenue.
2. Find average revenue.
3. Identify highest sales month.
4. Append new month sale = 20000.
5. Sort the list in descending order.

```python
sales = [12000, 15000, 11000, 18000, 17000]

# 1. Calculate total revenue
total = sum(sales)
print("Total revenue:", total)

# 2. Find average revenue
average = total / len(sales)
print("Average revenue:", average)

# 3. Identify highest sales
highest = max(sales)
print("Highest sales:", highest)

# 4. Append new month sale
new_sale = 20000
sales.append(new_sale)
print("After adding new sale:", sales)

# 5. Sort the list in descending order
descending_sales = sorted(sales, reverse=True)
print("Descending order:", descending_sales)
```

Question 9:You are building a simple sales analysis system.
Given:
products = ["Laptop", "Phone", "Tablet"]
prices = [50000, 20000, 15000]
quantities = [5, 10, 7]
Tasks:
1. Compute total revenue for each product.
2. Identify which product generated maximum revenue.
3. Store revenues in a new list.
4. Print a formatted summary like:

```python
# total revenue
products = ["Laptop", "Phone", "Tablet"]
prices = [50000, 20000, 15000]
quantities = [5, 10, 7]

revenues = []

for i in range(len(products)):
    revenue = prices[i] * quantities[i]
    revenues.append(revenue)

print(revenues)

# max revenue product
max_revenue = max(revenues)
max_revenue_product = products[revenues.index(max_revenue)]
print(max_revenue_product)


#3
new_revenue_list = revenues
print(new_revenue_list)

#4
for i in range(len(products)):
    print(f"{products[i]}: {revenues[i]}")




```
